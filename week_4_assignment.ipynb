```python
# Analyze data using SQL or Pandas.

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_units = df["Units Sold"].sum()
total_gross_sales = df["Gross Sales"].sum()
total_discounts = df["Discounts"].sum()
total_cogs = df["COGS"].sum()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Units Sold:", total_units)
print("Total Gross Sales:", total_gross_sales)
print("Total Discounts:", total_discounts)
print("Total COGS:", total_cogs)
```

```python
profit_margin = total_profit / total_sales

print("Overall Profit Margin:", round(profit_margin, 2))
```

```python
discount_rate = total_discounts / total_gross_sales

print("Overall Discount Rate:", round(discount_rate, 2))
```

```python
country_analysis = df.groupby("Country").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Units_Sold=("Units Sold", "sum")
).reset_index()

country_analysis
```

```python
country_analysis.sort_values(
    by="Sales",
    ascending=False
)
```

```python
country_analysis.sort_values(
    by="Profit",
    ascending=False
)
```

```python
country_analysis["Profit Margin"] = (
    country_analysis["Profit"] /
    country_analysis["Sales"]
)
```

```python
country_analysis.sort_values(
    by="Profit Margin",
    ascending=False
)
```

```python
product_analysis = df.groupby("Product").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Units_Sold=("Units Sold", "sum")
).reset_index()

product_analysis
```

```python
year_analysis = df.groupby("Year").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Units_Sold=("Units Sold", "sum")
).reset_index()

year_analysis
```

```python
year_analysis["Profit Margin"] = (
    year_analysis["Profit"] /
    year_analysis["Sales"]
)
```

```python
sales_growth = (
    (year_analysis["Sales"].iloc[-1] -
     year_analysis["Sales"].iloc[0])
    / year_analysis["Sales"].iloc[0]
) * 100

print("Sales Growth (%):", sales_growth)
```

```python
profit_growth = (
    (year_analysis["Profit"].iloc[-1] -
     year_analysis["Profit"].iloc[0])
    / year_analysis["Profit"].iloc[0]
) * 100

print("Profit Growth (%):", profit_growth)
```
